### Handling Outliers

```python
all_numeric    = df.select_dtypes(include=np.number).columns # Select all numeric columns
feature_cols = all_numeric.drop(['quality'])  # features only
n_feat         = len(feature_cols) # number of features

# 2) Create a two‑column grid:  (scatter+box  |  histogram )  per feature
fig, axes = plt.subplots(n_feat, 2, figsize=(12, 3 * n_feat))

for i, col in enumerate(feature_cols):
    # --- Left panel : jittered scatter + boxplot --------------------
    sns.stripplot(x=np.ones(len(df)), y=df[col],
                  color='grey', alpha=0.4, size=3, ax=axes[i, 0]) # jittered scatter
    sns.boxplot(y=df[col], color='red', fliersize=0, ax=axes[i, 0]) # boxplot
    axes[i, 0].set_title(f'{col}: scatter & box') # title of the plot
    axes[i, 0].set_ylabel(col) # label for y-axis
    axes[i, 0].set_xlabel('') # no label for x-axis (just a dummy value)

    # --- Right panel : histogram -----------------------------------
    sns.histplot(df[col], bins=30, color='salmon', ax=axes[i, 1]) # histogram
    axes[i, 1].set_title(f'{col}: distribution') # title of the plot
    axes[i, 1].set_xlabel(col) # label for x-axis
    axes[i, 1].set_ylabel('count') # label for y-axis

plt.tight_layout() # Adjust layout to prevent overlap
plt.show()
```

```python
# Compute Tukey fences & build a mask that keeps rows *inside* the fence
Q1  = df[feature_cols].quantile(0.25)
Q3  = df[feature_cols].quantile(0.75)
IQR = Q3 - Q1

# A row is *kept* only if *all* its predictor values fall within [Q1‑1.5×IQR , Q3+1.5×IQR]
mask = ~((df[feature_cols] < (Q1 - 1.5 * IQR)) |
         (df[feature_cols] > (Q3 + 1.5 * IQR))).any(axis=1)

rows_before = len(df) # Count rows before outlier removal
print('Rows before outlier removal: ', rows_before)

df = df[mask].reset_index(drop=True) # Apply the mask to the DataFrame and reset the index

rows_after = len(df) # Count rows after outlier removal
print('Rows after  outlier removal: ', rows_after)
print('Rows containing outliers removed: ', (rows_before - rows_after)) # Count rows removed and print it
```
